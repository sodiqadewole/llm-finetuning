### Training Reward Models

Reward Models (RMs) are typically trained using supervised learning on datasets containing pairs of preferred and non-preferred responses. We aim to learn a function that assigns higher scores to preferred responses, enabling the model to rank outputs based on preferences.

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
from trl import RewardTrainer
from datasets import load_dataset

trainer = RewardTrainer(
    model="Qwen/Qwen3-0.6B",
    train_dataset=load_dataset("trl-lib/ultrafeedback_binarized", split="train"),
)
```

```python
trainer.train()
```
